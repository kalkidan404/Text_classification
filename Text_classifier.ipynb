```python
import re

def clean_text(text):
    text = re.sub(r"<br\s*/?>", " ", text)
    text = text.lower()
    return text
```

```python
all_tokens = []

for review in dataset["train"]:
    text = clean_text(review["text"])
    tokens = tokenize(text)
    all_tokens.extend(tokens)

print("Total tokens:", len(all_tokens))
print("Unique tokens:", len(set(all_tokens)))
```

```python
from collections import Counter

word_counts = Counter(all_tokens)

print(word_counts.most_common(20))
```

```python
vocab_size = 20000

most_common_words = word_counts.most_common(vocab_size - 2)

word_to_idx = {
    "<PAD>": 0,
    "<UNK>": 1
}

for idx, (word, count) in enumerate(most_common_words, start=2):
    word_to_idx[word] = idx

print("Vocabulary size:", len(word_to_idx))
print("Movie ID:", word_to_idx["movie"])
print("Not ID:", word_to_idx["not"])
print("Unknown ID:", word_to_idx["<UNK>"])
print("Padding ID:", word_to_idx["<PAD>"])
```

```python
def encode_text(text):
    tokens = tokenize(clean_text(text))

    return [
        word_to_idx.get(token, word_to_idx["<UNK>"])
        for token in tokens
    ]
```

```python
#encoding it
sample_review = dataset["train"][0]["text"]

encoded_review = encode_text(sample_review)

print("Original:", sample_review[:200])
print("Encoded:", encoded_review[:30])
print("Number of tokens:", len(encoded_review))
```

```python
MAX_LENGTH = 256

def encode_and_pad(text):
    encoded = encode_text(text)

    if len(encoded) < MAX_LENGTH:
        encoded += [word_to_idx["<PAD>"]] * (MAX_LENGTH - len(encoded))
    else:
        encoded = encoded[:MAX_LENGTH]

    return encoded
```

```python
padded_review = encode_and_pad(sample_review)

print("Original length:", len(encoded_review))
print("Processed length:", len(padded_review))
print("First 30 values:", padded_review[:30])
print("Last 10 values:", padded_review[-10:])
```

```python
short_review = "This movie was absolutely fantastic."

short_encoded = encode_and_pad(short_review)

print("Length:", len(short_encoded))
print("Encoded:", short_encoded)
print("Padding count:", short_encoded.count(0))
```

```python
import torch

X_train = [
    encode_and_pad(review["text"])
    for review in dataset["train"]
]

y_train = dataset["train"]["label"]

X_test = [
    encode_and_pad(review["text"])
    for review in dataset["test"]
]

y_test = dataset["test"]["label"]
```

```python
X_train = torch.tensor(X_train, dtype=torch.long)
y_train = torch.tensor(y_train, dtype=torch.long)

X_test = torch.tensor(X_test, dtype=torch.long)
y_test = torch.tensor(y_test, dtype=torch.long)
```

```python
print("Training inputs:", X_train.shape)
print("Training labels:", y_train.shape)

print("Testing inputs:", X_test.shape)
print("Testing labels:", y_test.shape)
```

creating a validation data by splitting


```python
from torch.utils.data import TensorDataset, DataLoader, random_split
```

```python
train_dataset = TensorDataset(X_train, y_train)
test_dataset = TensorDataset(X_test, y_test)
```

```python
train_size = 20000
val_size = 5000

train_subset, val_subset = random_split(
    train_dataset,
    [train_size, val_size],
    generator=torch.Generator().manual_seed(42)
)
```

```python
class SentimentClassifier(nn.Module):
    def __init__(self):
        super().__init__()

        self.embedding = nn.Embedding(
            num_embeddings=20000,
            embedding_dim=100,
            padding_idx=0
        )

        self.classifier = nn.Sequential(
            nn.Linear(100, 64),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(64, 2)
        )

    def forward(self, x):
        # Convert word IDs into embeddings
        embedded = self.embedding(x)

        # Create padding mask
        mask = (x != 0).unsqueeze(-1)

        # Exclude padding
        embedded = embedded * mask

        # Mean pooling
        summed = embedded.sum(dim=1)
        lengths = mask.sum(dim=1).clamp(min=1)

        pooled = summed / lengths

        # Classification
        output = self.classifier(pooled)

        return output
```

```python
loss_fn = nn.CrossEntropyLoss()
```

```python
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)
```

```python
def train_one_epoch(model, train_loader, loss_fn, optimizer):
    model.train()

    total_loss = 0
    correct = 0
    total = 0

    for reviews, labels in train_loader:

        # Forward pass
        outputs = model(reviews)

        # Calculate loss
        loss = loss_fn(outputs, labels)

        # Clear previous gradients
        optimizer.zero_grad()

        # Backpropagation
        loss.backward()

        # Update parameters
        optimizer.step()

        # Track loss
        total_loss += loss.item()

        # Track accuracy
        predictions = outputs.argmax(dim=1)
        correct += (predictions == labels).sum().item()
        total += labels.size(0)

    average_loss = total_loss / len(train_loader)
    accuracy = correct / total

    return average_loss, accuracy
```

```python
def evaluate(model, data_loader, loss_fn):
    model.eval()

    total_loss = 0
    correct = 0
    total = 0

    with torch.no_grad():

        for reviews, labels in data_loader:

            outputs = model(reviews)
            loss = loss_fn(outputs, labels)

            total_loss += loss.item()

            predictions = outputs.argmax(dim=1)
            correct += (predictions == labels).sum().item()
            total += labels.size(0)

    average_loss = total_loss / len(data_loader)
    accuracy = correct / total

    return average_loss, accuracy
```

```python
EPOCHS = 10

for epoch in range(EPOCHS):

    train_loss, train_acc = train_one_epoch(
        model, train_loader, loss_fn, optimizer
    )

    val_loss, val_acc = evaluate(
        model, val_loader, loss_fn
    )

    print(
        f"Epoch {epoch+1}/{EPOCHS} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_acc:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_acc:.4f}"
    )
```

```python
test_loss, test_acc = evaluate(
    model, test_loader, loss_fn
)

print(f"Test Loss: {test_loss:.4f}")
print(f"Test Accuracy: {test_acc:.4f}")
```

```python
import torch
import torch.nn as nn
from torch.nn.utils.rnn import pack_padded_sequence


class LSTMSentimentClassifier(nn.Module):

    def __init__(self):
        super().__init__()

        self.embedding = nn.Embedding(
            num_embeddings=20000,
            embedding_dim=100,
            padding_idx=0
        )

        self.lstm = nn.LSTM(
            input_size=100,
            hidden_size=128,
            batch_first=True
        )

        self.dropout = nn.Dropout(0.3)

        self.classifier = nn.Linear(128, 2)

    def forward(self, x):

        lengths = (x != 0).sum(dim=1)

        embedded = self.embedding(x)

        packed = pack_padded_sequence(
            embedded,
            lengths.cpu(),
            batch_first=True,
            enforce_sorted=False
        )

        _, (hidden, cell) = self.lstm(packed)

        final_hidden = hidden[-1]

        output = self.dropout(final_hidden)

        output = self.classifier(output)

        return output
```
